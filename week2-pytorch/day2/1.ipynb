```python
import torch
```

reshape():  前后元素的总数要一样
view()：要求 Tensor 的内存布局满足条件

```python
x = torch.arange(12)
print(x)
print(x.reshape(3, 4))
print(x.view(3, 4))
print(x)
```

flatten：展平维度

```python
x=torch.randn(2,3,4)
print(x)
print(x.flatten().shape)
print(x.flatten(start_dim=1).shape)
```

unsqueeze：增加维度,squeeze:压缩维度

```python
x = torch.randn(1, 3, 224, 224)
print(x.shape)
x = x.squeeze(0)
print(x.shape)
```

cat：拼接 Tensor,,stack：新增一个维度后堆叠   
cat 是沿已有维度拼接，stack 是新增维度堆叠。

```python
a = torch.tensor([[1, 2]])
b = torch.tensor([[3, 4]])
x = torch.cat([a, b], dim=0)
print(x)
print(torch.cat([a,b], dim=1))
print(x.shape)
y = torch.stack([a, b])
print(y)
```
