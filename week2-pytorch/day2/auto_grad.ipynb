```python
#定义数据
x = torch .tensor([[10.0]])
y= torch.tensor([[3.0]])
print(x)
print(y)
```

```python
w = torch.rand(1,1,requires_grad=True)
b = torch.rand(1,1,requires_grad=True)
```

```python
#前向传播
z = x * y + w * b
print(z)
```

叶子节点和非叶子1节点，根据表达式而来

```python
#设置损失函数
loss = torch.nn.MSELoss()
loss_value =loss(z,y)
print(loss_value)
```

```python
#反向传播
loss_value.backward()
```

```python
#查看梯度
print(w.grad)
print(b.grad)
```

梯度累积

```python
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2
y.backward()
print("第一次:", x.grad)
z = x ** 3
z.backward()
print("第二次:", x.grad)
```

```python
#模拟神经网络 y=w*x
import torch

x = torch.tensor(2.0)
target = torch.tensor(10.0)
w = torch.tensor(1.0,requires_grad=True)

learning_rate = 0.1

for step in range(10):
    pred = w*x
    loss = (pred-target)**2
    loss.backward()
    with torch.no_grad():#只是计算不需要计算梯度
        w-=learning_rate*w.grad

    w.grad.zero_()#每一轮梯度要归零，不然就会叠加

    print(f"step={step},"f"w={w.item():4f}",f"loss={loss.item():4f}")#item,把标量转化为数字
```
