神经网络模块 sigmoid函数[0,1], Tanh函数[-1,1],ReLU函数(0 or x),Softmax函数
torch.nn

```python
import torch 
import torch.nn as nn

linear = nn.Linear(5,2)

nn.init.zeros_(linear.weight)#初始化参数有转置的效果
print(linear.weight)
```

```python
nn.init.ones_(linear.weight)
print(linear.weight)
nn.init.constant_()#常数
nn.init.eye_()#对角线
```

```python
#正态分布初始化
nn.init.normal_(linear.weight,mean=0,std=0.1)
print(linear.weight)
```

```python
#Xavier初始化(根据输入与输出)（1正态2.均匀）
nn.init.xavier_normal_(linear.weight)
nn.init.xavier_uniform_(linear.weight)
#He初始化，根据输入神经元
```
