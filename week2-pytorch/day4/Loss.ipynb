损失函数，评测估计好不好的问题
1.分类任务a.二元交叉熵损失函数（nn.BCELoss()）b.多分类任务（nn.CrossEntropyLoss()）

```python
import torch
import torch.nn as nn
```

```python
input = torch.randn(3,2)
print(input)
#得到预测值
pred = torch.sigmoid(input)
```

```python
target  = torch.tensor([[0,1],[1,0],[0,1]],dtype=torch.float)
print(target)
```

```python
loss = nn.BCELoss()
print(loss(pred,target))
```

b.多分类任务（nn.CrossEntropyLoss()）

```python
x = torch.randn(4,6)
print(x)
```

```python
target = torch.tensor([1,0,3,0])#有的可能是独热编码
loss = nn.CrossEntropyLoss()
print(loss(x,target))
```
