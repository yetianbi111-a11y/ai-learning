Dataset 与 DataLoader

Dataset 可以简单理解成：

负责告诉 PyTorch：数据在哪里，以及如何取出第 i 个样本。
三个函数__init__()，__len__(),__getitem__()

```python
class MyDataset(torch.utils.data.Dataset):
    def __init__(self,X,y):
        self.X=X
        self.y=y

    def __len__(self):
        return len(self.X)

    def __getitem__(self, index):
        return self.X[index],self.y[index]

X = torch.randn(100,10)
y = torch.randint(0,4,(100,))

dataset = MyDataset(X,y)

print(len(dataset))
```

```python
x,y = dataset[0]
print(x)
print(x.shape)
print(y)
print(y.shape)
```

DataLoader 通俗理解就是Dataset相当于一大堆资料，使用loader一批次一批次送入网络（全部同时送入数据太多，同时计算量很大）

```python
from torch.utils.data import DataLoader

loader = DataLoader(
    dataset,
    batch_size=16,
    shuffle=True#每个epoch打乱顺序
)

for x,y in loader:
    print(x.shape)
    print(y.shape)
```
